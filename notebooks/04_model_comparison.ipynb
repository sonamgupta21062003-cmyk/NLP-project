```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
comparison = pd.DataFrame({

    "Model":[
        "Logistic Regression",
        "Linear SVM"
    ],

    "Accuracy":[
        0.9777,
        0.9808
    ],

    "Vectorizer":[
        "TF-IDF",
        "TF-IDF"
    ],

    "Training Speed":[
        "Fast",
        "Fast"
    ],

    "Prediction Speed":[
        "Very Fast",
        "Very Fast"
    ],

    "Memory":[
        "Low",
        "Low"
    ]

})

comparison
```

```python
plt.figure(figsize=(6,4))

plt.bar(
    comparison["Model"],
    comparison["Accuracy"]
)

plt.ylim(0.95,1.00)

plt.ylabel("Accuracy")

plt.title("Model Comparison")

plt.show()
```

```python
best = comparison.sort_values(
    by="Accuracy",
    ascending=False
)

best
```

```python
comparison.to_csv(
    "../reports/metrics/model_comparison.csv",
    index=False
)
```
