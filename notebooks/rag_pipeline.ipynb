```python
import os
import sys
import pandas as pd

# 1. Point to the backend directory level safely
sys.path.append(os.path.abspath(os.path.join('..', 'backend')))

# 2. Let's create a functional reader that scans your directory structure directly
def read_local_policy(category: str) -> str:
    """
    Reads policy text documentation directly from your backend/knowledge_base/ subfolders.
    """
    base_path = os.path.join('..', 'backend', 'knowledge_base', category)
    
    if not os.path.exists(base_path):
        return "Thank you for contacting customer support. A live agent will look into your request shortly."
        
    # Read the files inside that directory (assuming standard text files exist inside them)
    try:
        files = [f for f in os.listdir(base_path) if os.path.isfile(os.path.join(base_path, f))]
        if not files:
            return "Standard corporate guidelines apply."
            
        # Read the first available file content as text documentation context
        with open(os.path.join(base_path, files[0]), 'r', encoding='utf-8') as file_reader:
            return file_reader.read()
    except Exception:
        return "Standard corporate guidelines apply."

print("RAG Prototyping environment successfully aligned with your folder structure!")
```

```python
def simulate_rag_pipeline(user_query: str, predicted_intent: str) -> str:
    """
    Simulates the RAG context injection using your actual folder layouts.
    """
    # Map intent strings cleanly to your visible folders
    intent_to_folder = {
        "check_refund_policy": "refunds",
        "get_refund": "refunds",
        "change_shipping_address": "shipping",
        "delivery_options": "shipping",
        "check_invoice": "payments",
        "payment_issue": "payments",
        "create_account": "accounts",
        "edit_account": "accounts"
    }
    
    # Identify folder target or use default fallback string
    target_category = intent_to_folder.get(predicted_intent, "faq")
    retrieved_policy = read_local_policy(target_category)
    
    prompt_template = f"""
    [SYSTEM INSTRUCTION]: You are an elite customer service AI. Answer the user's question using ONLY the provided policy context.
    
    [GROUNDING CONTEXT FROM YOUR DIRECTORY]:
    {retrieved_policy}
    
    [USER QUERY]: 
    {user_query}
    """
    return prompt_template
```

```python
# Test a pipeline execution simulation run
test_query = "Can I get my money back for this broken order?"
test_intent = "check_refund_policy"

assembled_prompt = simulate_rag_pipeline(test_query, test_intent)
print("=== ASSEMBLED CONTEXT PROMPT ===")
print(assembled_prompt)
```
